# Cas Kaggle: Resolem un problema d’aprenentatge computacional

## Objectius

Els objectius d’aquest cas són:

* Aplicar de manera autònoma tot el que heu après al llarg del curs (dades, models, validació, mètriques i *data leakage*) a una base de dades que trieu vosaltres.

* Plantejar el problema: decidir quina pregunta es respon, amb quines dades, amb quina mètrica i amb quin protocol de validació.

* Comparar diferents mètodes i trobar-ne els millors hiperparàmetres amb un protocol honest.

* Analitzar els resultats i defensar les decisions preses amb evidència, risc i límit.

* Documentar el treball en un repositori de GitHub que qualsevol persona pugui reproduir.

A la Pràctica 1 les preguntes estaven pautades; aquí no. L’enunciat us diu què s’avalua i quin és el mínim que s’espera, i l’estructura del projecte la decidiu vosaltres.

## Bases de dades

Podeu triar una base de dades del fitxer `kaggle.xlsx` (a Caronte, amb la dificultat, la versatilitat i la tasca de cada una) o qualsevol altra de [Kaggle](https://www.kaggle.com/search?q=machine+learning). La llista és una ajuda, no una restricció. La tasca pot ser de regressió, de classificació o de clustering.

Una base de dades massa petita o massa fàcil no permet comparar models, i una massa gran no es pot treballar amb els vostres ordinadors. No podeu fer servir les que ja hem treballat al curs (Titanic, Ames House Prices i la de diabetis de Pima). Cal indicar la base de dades triada (l’enllaç de Kaggle) a la tasca «Base de dades del cas Kaggle» de Caronte, abans de la sessió 3 del 23 de novembre.

## Treball en grup

El cas es durà a terme en grups d’1 o 2 persones.

El curs dedica sis sessions al cas:

* 10 de novembre: sessió 1 informativa (introducció, selecció de bases de dades, llibreries i GitHub).
* 16 de novembre: sessió 2 de treball, hands-on «Com plantejar un problema».
* 23 de novembre: sessió 3 de treball (data mining i anàlisi d’atributs).

* 30 de novembre: sessió 5 de treball (comparativa de mètodes).
* 15 de desembre: sessió 6 d’avaluació (presentació opcional).

Pel que fa a l’entrega, caldrà lliurar a través de Caronte l’enllaç al vostre repositori de GitHub a la tasca «Entrega Casos Kaggle», que s’obre l’1 de desembre i es tanca el dilluns 14 de desembre a les 00:05, és a dir, la nit del diumenge 13 al dilluns 14. El codi i l’explicació del codi i dels resultats que hi hagi al repositori són el que s’avalua. El cas val el 40% de la nota final de l’assignatura i no és recuperable.

*Nota*: les sessions de treball són per resoldre dubtes. És molt recomanable arribar-hi amb el treball avançat.

# Continguts que s’avaluaran en el cas

1. **Introducció a la base de dades i anàlisi dels atributs i de les correlacions**, 2.5 punts

* Origen de les dades, tasca i atribut a predir.

* Tipus d’atributs, valors perduts i distribució del target.

* Relacions entre atributs i amb el target.

2. **Mètode**, 2.5 punts

* Mètrica, models, validació creuada i cerca d’hiperparàmetres.

* Comparativa de mètodes.

3. **Descripció dels resultats**, 3 punts

* Matrius de confusió, gràfiques dels models i de les dades, corbes ROC o equivalents.

* Exemples d’errors (falsos positius i falsos negatius) i anàlisi per grups.

4. **Repositori de GitHub**, 1 punt

* Estructura, README i reproductibilitat.

5. **Presentació oral**, 1 punt

* Opcional. Qui no presenti té un 0 en aquest apartat.

Aquests cinc apartats sumen 10 i determinen la nota del cas. Cal tenir en compte que les puntuacions dins de cada apartat són orientatives. La presentació oral és opcional: sense fer-la es pot arribar a un 9.0 sobre 10 al cas.

Per arribar a un 7.5 sobre 10 al cas cal complir el mínim de cada un dels quatre primers apartats, que trobareu a sota de cada títol. La resta és opcional: respondre-ho amb rigor i amb conclusions pròpies fa pujar la nota, i també la fan pujar la iniciativa i la independència.

**IMPORTANT**: El criteri principal d’avaluació serà la capacitat de mantenir una narrativa coherent a mesura que es presenten els resultats i de defensar cada decisió amb evidència. A les sessions de treball us plantejarem preguntes guia per a cada apartat. No es penalitzarà deixar-ne alguna sense respondre, sempre que el procés seguit i les conclusions extretes siguin consistents i ben fonamentats.

Les figures són una part central de l’avaluació, no una il·lustració. Cada decisió important (la distribució del target, els valors perdut, les correlacions, la comparació de models, el resultat final i els errors) ha d’anar acompanyada d’una figura llegible, amb títol, eixos amb nom i llegenda, i d’un text que digui què se’n dedueix. Una figura sense interpretació val poc, i un resultat clau que només surt en una taula o en una frase perd força.

## 2.1 Introducció a la base de dades i anàlisi dels atributs i de les correlacions, **2.5 punts**

### Introducció a la base de dades

Abans de començar, cal que el lector entengui de quines dades parteu i què es vol predir.

**Mínim per al 7.5:** presentar les dades, la pregunta que es vol respondre i l’atribut a predir.

La idea es predir preus d'allotjament

### Anàlisi dels atributs i de les correlacions

Analitzeu les dades amb figures i escriviu què en deduïu.

**Mínim per al 7.5:** descriure els atributs, decidir què significa cada valor perdut i mirar la distribució del target i les correlacions amb figures.

Tot el que s’aprèn de les dades (medianes, modes, escalats, codificacions, selecció d’atributs, remostratge) s’ha d’ajustar només amb el conjunt d’entrenament, dins d’un Pipeline. Partiu abans d’ajustar res. Si les dades tenen temps o grups (per exemple, la mateixa persona en diverses files), la partició ha de respectar-ho. Consulteu `data_leakage.pdf`.

## 2.2 Mètode, **2.5 punts**

Cal escollir una sèrie de models i, mitjançant validació creuada, seleccionar el millor amb els seus hiperparàmetres òptims. Expliqueu breument què fa cada model, sobretot els que no s’han vist a classe, i per què l’heu triat.

**Mínim per al 7.5:** calcular una baseline, triar i justificar la mètrica, comparar com a mínim quatre models amb validació creuada i fer una cerca d’hiperparàmetres per a cadascun, tot ajustat només amb train i dins d’un Pipeline.

El conjunt de test es reserva per al final: només s’utilitza una vegada, amb el model ja triat, i cap decisió (model, hiperparàmetres, mètrica, llindar) es pren mirant el test.

## 2.3 Descripció dels resultats, **3 punts**

Un cop seleccionat el model final, cal elaborar un report amb els resultats.

**Mínim per al 7.5:** indicar quin és el model final i amb quin criteri l’heu triat abans de mirar el test, avaluar-lo al test una sola vegada, mostrar les gràfiques que corresponguin a la vostra tasca i explicar un cas d’ús i possibles millores.

## 2.4 Repositori de GitHub, **1 punt**

El repositori és el lliurament: qualsevol persona ha de poder entendre què heu fet i reproduir-ho. A `github.pdf` trobareu com crear-lo i com treballar amb Git.

**Mínim per al 7.5:** un repositori que es pugui executar d’extrem a extrem, amb un README.

**Què ha de tenir:**

-  Un `README.md` amb la pregunta, la base de dades (amb l’enllaç a Kaggle), els resultats principals i com executar el projecte.
-  Un o més notebooks que s’executin d’extrem a extrem, amb text que expliqui cada decisió i el resultat que se n’ha derivat.
-  Un fitxer amb les llibreries necessàries i una llavor (*seed*) fixada allà on hi hagi aleatorietat.
-  Les dades només si són petites; si no, l’enllaç per descarregar-les.
-  Un historial de *commits* en què es vegi la feina de cada membre del grup.
-  L’usuari `silvanasp` afegit com a col·laborador del repositori (a GitHub, *Settings*, *Collaborators*), perquè hi pugui accedir encara que sigui privat.

Una organització possible és:

```
projecte-ml/
|-- data/
|-- notebooks/
|   |-- 01_exploration.ipynb
|   `-- 02_models.ipynb
|-- src/
`-- README.md
```

Podeu trobar exemples de notebooks aplicats a casos concrets a [datauab.github.io](https://datauab.github.io/).
